```python
dtypes=df.dtypes
n_unique = df.nunique()
pd.DataFrame({"Dtype":dtypes,"Num_unique":n_unique}).T

```

```python
null = df.isnull().sum()
ratio = null/df.shape[0]
pd.DataFrame({"Null":null,"Ratio":ratio})
```

```python
df = df.dropna(subset=['start_station_id', 'end_station_id','member_birth_year'])
```

```python
mode = df['member_gender'].mode()[0]
df['member_gender'] = df['member_gender'].fillna(mode)
```

```python
null = df.isnull().sum()
ratio = null/df.shape[0]
pd.DataFrame({"Null":null,"Ratio":ratio})
```

```python
df['user_type'] = df['user_type'].str.strip()
```

```python
df['member_gender'] = df['member_gender'].str.strip()
```

```python
df['age'] = 2019 - df['member_birth_year']
```

```python
df = df[(df['age'] >= 18) & (df['age'] <= 80)]
```

```python
df['age'].describe()
```

```python
df['duration_sec'].describe()
```

```python
df.drop_duplicates(inplace=True)
```

```python
Q1 = df['duration_sec'].quantile(0.25) 
Q3 = df['duration_sec'].quantile(0.75) 
IQR = Q3 - Q1 

Lower_Fence = Q1 - 1.5 * IQR 
Upper_Fence = Q3 + 1.5 * IQR 
Lower_Outliers = df[df['duration_sec'] < Lower_Fence]['duration_sec'].values 
Upper_Outliers = df[df['duration_sec'] > Upper_Fence]['duration_sec'].values 

df = df[(df['duration_sec'] >= Lower_Fence) &(df['duration_sec'] <= Upper_Fence)]
```

```python
df['duration_sec'].describe()
```

```python
df['duration_min'] = df['duration_sec'] / 60
```

```python
df['duration_min'].describe()
```

```python
df["start_time"] = pd.to_datetime(df["start_time"], errors="coerce")
df = df.dropna(subset=["start_time"])
```

```python
df["hour"] = df["start_time"].dt.hour
```

```python
hour_data = (
    df.groupby("hour")
    .size()
    .reset_index(name="trips")
    .sort_values("hour")
)
```

```python
bins = [18, 25, 35, 45, 60, 80]
labels = ['18-25', '26-35', '36-45', '46-60', '61-80']

df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
```

```python
df = pd.get_dummies(
    df,
    columns=[
        'bike_share_for_all_trip'
    ],
    drop_first=True
)
```

```python
import plotly.express as px
import plotly.graph_objects as go
px.defaults.template = "plotly_white"

# Color palette
COLORS = {
    'primary': '#00BCD4',
    'secondary': '#FF9800', 
    'accent': '#9C27B0',
    'light_blue': '#4FC3F7',
    'dark': '#37474F'
}


```

```python
hours = hour_data["hour"]
trips_hour = hour_data["trips"]

fig1 = go.Figure()

fig1.add_trace(go.Scatter(
    x=hours,
    y=trips_hour,
    mode="lines+markers",
    line=dict(width=3, color=COLORS["accent"]),
    marker=dict(size=6),
    hovertemplate="Hour: %{x}<br>Trips: %{y}<extra></extra>"
))

fig1.update_layout(
    title=dict(text="Trips by Hour of Day", font=dict(size=18, color=COLORS["dark"])),
    template="plotly_white",
    height=350,

    font=dict(family="Segoe UI, Arial", size=12),
    xaxis=dict(
        title="Hour of Day",
        title_font=dict(size=16),
        dtick=1,
        showgrid=False,
        linecolor="lightgray"
        
    ),
    yaxis=dict(
        title="Number of Trips",
        title_font=dict(size=16),
        showgrid=True,
        gridcolor="rgba(0,0,0,0.05)",
        linecolor="lightgray"
    ),
    margin=dict(l=50, r=30, t=50, b=40)
)

fig1.show()
```

```python
user_counts = df['user_type'].value_counts()

fig3 = go.Figure(data=[go.Pie(
    labels=user_counts.index,
    values=user_counts.values,
    hole=0.65,
    marker=dict(
        colors=[COLORS['primary'], COLORS['secondary']],
        line=dict(width=2, color='white')
    ),
    textinfo='none',
    hoverinfo='label+percent'
)])

user_counts = df['user_type'].value_counts()
total = user_counts.sum()

max_label = user_counts.idxmax()
max_percent = (user_counts.max() / total) * 100
fig3.update_layout(
    title=dict(text='Subscriber vs Customer Usage', font=dict(size=24, color=COLORS['dark'])),
    template='plotly_white',
    height=350,
    width=500,
    font=dict(family="Segoe UI, Arial", size=12),
    paper_bgcolor='white',
    showlegend=True,
    legend=dict(orientation="h", yanchor="bottom", y=-0.15, x=0.5, xanchor="center"),
    margin=dict(l=40, r=40, t=50, b=60),
    annotations=[
        dict(
            text=f"{max_percent:.1f}%",
            x=0.5,
            y=0.5,
            font_size=28,
            showarrow=False,
            font_family="Segoe UI"
        ),
        dict(
            text=max_label,
            x=0.5,
            y=0.35,
            font_size=14,
            showarrow=False,
            font_color='gray'
        )
    ]
)
```

```python
gender_counts = df['member_gender'].value_counts()

gender_percent = (gender_counts / gender_counts.sum()) * 100

order = ['Male', 'Female', 'Other']
gender_counts = gender_counts.reindex(order).fillna(0)
gender_percent = gender_percent.reindex(order).fillna(0)

fig4 = go.Figure()

fig4.add_trace(go.Bar(
    x=gender_counts.index,
    y=gender_counts.values,
    marker=dict(
        color=[COLORS['primary'], COLORS['secondary'], COLORS['accent']],
        line=dict(width=1, color='white')
    ),
    text=[f"{p:.1f}%" for p in gender_percent.values],
    textposition='none',
    textfont=dict(size=12, color=COLORS['dark'])
))

fig4.update_layout(
    title=dict(text='Trips By Gender', font=dict(size=18, color=COLORS['dark'])),
    template='plotly_white',
    height=400,
    width =500,
    font=dict(family="Segoe UI, Arial", size=12),
    paper_bgcolor='white',
    plot_bgcolor='white',
    xaxis=dict(showgrid=False, linecolor='lightgray'),
    yaxis=dict(showgrid=True, gridcolor='rgba(0,0,0,0.05)', linecolor='lightgray'),
    margin=dict(l=40, r=30, t=50, b=40),
    bargap=0.3
)

fig4.show()
```

```python
top_stations = (
    df['start_station_name']
    .value_counts()
    .head(10)
    .sort_values(ascending=False)  
    )
stations = top_stations.index
counts = top_stations.values

colors_stations = [
    COLORS['primary'] if value in top_stations.head(3).values
    else COLORS['light_blue']
    for value in counts
]

fig6 = go.Figure()

fig6.add_trace(go.Bar(
    y=stations,
    x=counts,
    orientation='h',
    marker=dict(color=colors_stations, line=dict(width=1, color='white')),
    text=[f'{c:,}' for c in counts],
    textposition='inside',
    textfont=dict(size=10, color=COLORS['dark'])
))
fig6.update_layout(
    title=dict(text='Top 10 Stations', font=dict(size=24, color=COLORS['dark'])),
    template='plotly_white',
    height=400,
    font=dict(family="Segoe UI, Arial", size=11),
    paper_bgcolor='white',
    plot_bgcolor='white',
    xaxis=dict(showgrid=True, gridcolor='rgba(0,0,0,0.05)', linecolor='lightgray', title='Number of Trips'),
    yaxis=dict(showgrid=False, linecolor='lightgray', autorange="reversed"),
    margin=dict(l=120, r=60, t=50, b=40),
    bargap=0.2
)
```
