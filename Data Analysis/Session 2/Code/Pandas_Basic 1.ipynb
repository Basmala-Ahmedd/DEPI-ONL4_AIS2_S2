```python
data = [[1,444,"adc"],
        [2,555,"def"],
        [3,666,"ghi"],
        [4,np.nan,"xyz"],
        ]

df = pd.DataFrame(data,columns=["ID","Std_id","Dept"])
df
```

```python
null = df.isnull().sum()
pd.DataFrame(null).T
```

```python
def isNUll(df):
    null = df.isnull().sum()  
    return pd.DataFrame(null).T
```
