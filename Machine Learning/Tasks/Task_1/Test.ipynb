```python
import numpy as np
import matplotlib.pyplot as plt
from Linear_Model import LinearModel_Basmala
```

```python
X = np.array([[1],[2],[3],[4]])
y = np.array([2,4,6,8])
```

```python
model_linear = LinearModel_Basmala(
    alpha=0.01,
    num_iters=1000,
    degree=1,
    regularization=None
)

model_linear.fit(X, y)
model_linear.plot_training(X, y)
plt.show()

print("Linear MSE:", model_linear.mse(X, y))
```

```python
model_ridge = LinearModel_Basmala(
    alpha=0.01,
    num_iters=2000,
    degree=1,
    regularization="ridge",
    lambda_=1
)

model_ridge.fit(X, y)
model_ridge.plot_training(X, y)
plt.show()

print("Ridge MSE:", model_ridge.mse(X, y))
```

```python
model_lasso = LinearModel_Basmala(
    alpha=0.01,
    num_iters=5000,
    degree = 1,
    regularization="lasso",
    lambda_= 15
)

model_lasso.fit(X, y)
model_lasso.plot_training(X, y)
plt.show()

print("Lasso MSE:", model_lasso.mse(X, y))
model_lasso.mse(X,y)
```

```python
model_poly = LinearModel_Basmala(
    alpha=0.001,
    num_iters=3000,
    degree=2,
    regularization=None
)

model_poly.fit(X, y)
model_poly.plot_training(X, y)
plt.show()

print("Polynomial MSE:", model_poly.mse(X, y))
```
