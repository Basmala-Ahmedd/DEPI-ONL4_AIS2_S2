```python
import numpy as np
import matplotlib.pyplot as plt
from Linear_Regression import LinearRegression_Basmala
```

```python
X = np.array([50, 60, 70, 80, 90])
y = np.array([150, 180, 210, 240, 270])
```

```python
# model = LinearRegression_Basmala(alpha=0.001, num_iters=100)
model = LinearRegression_Basmala(alpha=0.0001, num_iters=100) # for optimal solution.
model.fit(X, y)
```

```python
print("theta_0:", model.theta_0)
print("theta_1:", model.theta_1)
```

```python
predicted_price = model.predict(70)
print("Predicted price for 70 m²:", predicted_price)
```

```python
model.plot_training(X,y) # The model diverged due to a large learning rate. alpha=0.001
```

```python
model_large = LinearRegression_Basmala(alpha=0.1, num_iters=100)
model_large.fit(X, y)
print("theta_0:", model_large.theta_0)
print("theta_1:", model_large.theta_1)
predicted_price = model_large.predict(70)
print("Predicted price for 70 m²:", predicted_price)
model_large.plot_training(X,y)
```
